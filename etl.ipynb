**3. Validação de Qualidade e Camada Silver**

Nesta etapa, os dados da camada Bronze são limpos, padronizados e validados.

Principais tratamentos:
- Padronização dos nomes das colunas
- Conversão de tipos
- Tratamento de valores ausentes
- Remoção de duplicidades
- Validação de anos, UFs, municípios e percentuais
- Separação de registros inválidos em quarentena

```python
from pathlib import Path
from datetime import datetime
import pandas as pd
import unicodedata

BRONZE_PATH = Path("data/bronze")
SILVER_PATH = Path("data/silver")
QUARANTINE_PATH = SILVER_PATH / "quarantine"

SILVER_PATH.mkdir(parents=True, exist_ok=True)
QUARANTINE_PATH.mkdir(parents=True, exist_ok=True)

print("Diretórios da Silver criados com sucesso.")
```

Funções Auxiliares:

```python
def normalizar_nome_coluna(coluna: str) -> str:
    coluna = coluna.strip().lower()
    coluna = unicodedata.normalize("NFKD", coluna)
    coluna = coluna.encode("ascii", errors="ignore").decode("utf-8")
    coluna = coluna.replace(" ", "_")
    coluna = coluna.replace("-", "_")
    coluna = coluna.replace("/", "_")
    coluna = coluna.replace("(", "")
    coluna = coluna.replace(")", "")
    return coluna


def padronizar_colunas(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df.columns = [normalizar_nome_coluna(coluna) for coluna in df.columns]
    return df


def limpar_strings(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    for coluna in df.select_dtypes(include=["object", 'str']).columns:
        df[coluna] = (
            df[coluna]
            .astype("string")
            .str.strip()
            .replace({"nan": None, "None": None, "": None})
        )

    return df
```

Função para padronizar campos comuns:

```python
def padronizar_campos_comuns(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    if "ano" in df.columns:
        df["ano"] = pd.to_numeric(df["ano"], errors="coerce").astype("Int64")

    if "sigla_uf" in df.columns:
        df["sigla_uf"] = (
            df["sigla_uf"]
            .astype(str)
            .str.strip()
            .str.upper()
            .replace({"NAN": None, "NONE": None, "": None})
        )

    if "id_municipio" in df.columns:
        df["id_municipio"] = (
            df["id_municipio"]
            .astype(str)
            .str.replace(".0", "", regex=False)
            .str.strip()
            .str.zfill(7)
            .replace({"0000nan": None, "000None": None})
        )

    for coluna in df.columns:
        nome_coluna = coluna.lower()

        if (
            "percentual" in nome_coluna
            or "meta" in nome_coluna
            or "taxa" in nome_coluna
            or "indicador" in nome_coluna
        ):
            df[coluna] = pd.to_numeric(df[coluna], errors="coerce")

    df["_data_processamento"] = datetime.now()
    df["_camada_origem"] = "bronze"

    return df
```

Função para separar válidos e inválidos:

Essa função não apaga os dados ruins diretamente. Ela separa os registros inválidos em uma pasta de quarentena.

```python
def separar_validos_invalidos(df: pd.DataFrame, chaves_obrigatorias: list[str]):
    df = df.copy()
    condicao_invalida = pd.Series(False, index=df.index)

    for chave in chaves_obrigatorias:
        if chave in df.columns:
            condicao_invalida = condicao_invalida | df[chave].isna()
        else:
            print(f"Atenção: chave obrigatória não encontrada: {chave}")

    if "ano" in df.columns:
        condicao_invalida = condicao_invalida | df["ano"].isna()
        condicao_invalida = condicao_invalida | ~df["ano"].between(2023, 2030)

    if "sigla_uf" in df.columns:
        condicao_invalida = condicao_invalida | df["sigla_uf"].notna() & (df["sigla_uf"].str.len() != 2)

    if "id_municipio" in df.columns:
        condicao_invalida = condicao_invalida | df["id_municipio"].notna() & (df["id_municipio"].str.len() != 7)

    for coluna in df.columns:
        nome_coluna = coluna.lower()

        if (
            "percentual" in nome_coluna
            or "meta" in nome_coluna
            or "taxa" in nome_coluna
            or "indicador" in nome_coluna
        ):
            condicao_invalida = condicao_invalida | (
                df[coluna].notna() & ~df[coluna].between(0, 100)
            )

    validos = df[~condicao_invalida].copy()
    invalidos = df[condicao_invalida].copy()

    validos["_status_validacao"] = "valido"

    if not invalidos.empty:
        invalidos["_status_validacao"] = "invalido"

    return validos, invalidos
```

Função principal para processar uma tabela Bronze:

```python
def processar_tabela_silver(
    nome_arquivo: str,
    chaves_obrigatorias: list[str],
    chave_deduplicacao: list[str]
):
    caminho_entrada = BRONZE_PATH / nome_arquivo

    if not caminho_entrada.exists():
        print(f"Arquivo não encontrado: {caminho_entrada}")
        return None

    df = pd.read_parquet(caminho_entrada)

    total_bronze = len(df)

    df = padronizar_colunas(df)
    df = limpar_strings(df)
    df = padronizar_campos_comuns(df)

    duplicados = pd.DataFrame()

    colunas_dedup_existentes = [
        coluna for coluna in chave_deduplicacao if coluna in df.columns
    ]

    if colunas_dedup_existentes:
        duplicados = df[df.duplicated(subset=colunas_dedup_existentes, keep=False)].copy()
        df = df.drop_duplicates(subset=colunas_dedup_existentes, keep="last")

    validos, invalidos = separar_validos_invalidos(df, chaves_obrigatorias)

    nome_base = nome_arquivo.replace(".parquet", "")
    caminho_saida = SILVER_PATH / f"silver_{nome_base}.parquet"

    validos.to_parquet(caminho_saida, index=False)

    if not invalidos.empty:
        invalidos.to_parquet(
            QUARANTINE_PATH / f"invalidos_{nome_base}.parquet",
            index=False
        )

    if not duplicados.empty:
        duplicados.to_parquet(
            QUARANTINE_PATH / f"duplicados_{nome_base}.parquet",
            index=False
        )

    resumo = {
        "arquivo": nome_arquivo,
        "total_bronze": total_bronze,
        "total_silver": len(validos),
        "total_invalidos": len(invalidos),
        "total_duplicados": len(duplicados),
        "saida": str(caminho_saida)
    }

    print(f"Processado: {nome_base}")
    print(f"Bronze: {total_bronze}")
    print(f"Silver válidos: {len(validos)}")
    print(f"Inválidos: {len(invalidos)}")
    print(f"Duplicados: {len(duplicados)}")
    print("-" * 50)

    return resumo
```

Criar uma tabela integrada municipal na Silver:

```python
#angelica
print("Etapa 1/9: lendo as bases Silver de metas e avaliacao municipais...")
meta_municipio = pd.read_parquet(
    SILVER_PATH / "silver_meta_alfabetizacao_municipio.parquet"
)
avaliacao_municipio = pd.read_parquet(
    SILVER_PATH / "silver_avaliacao_alfabetizacao_municipio.parquet"
)

print("Etapa 2/9: padronizando chaves de integracao (ano, id_municipio e rede)...")
mapa_rede = {
    "1": "federal",
    "2": "estadual",
    "3": "municipal",
    "4": "privada",
    "5": "publica",
    "6": "publica",
    "pública": "publica",
    "publica": "publica",
    "municipal": "municipal",
    "estadual": "estadual",
    "federal": "federal",
    "privada": "privada"
}

for df in [meta_municipio, avaliacao_municipio]:
    df["ano"] = pd.to_numeric(df["ano"], errors="coerce").astype("Int64")
    df["id_municipio"] = (
        df["id_municipio"]
        .astype("string")
        .str.replace(".0", "", regex=False)
        .str.strip()
        .str.zfill(7)
    )
    df["rede"] = (
        df["rede"]
        .astype("string")
        .str.strip()
        .str.lower()
        .replace(mapa_rede)
    )

print("Etapa 3/9: identificando colunas de meta por ano...")
colunas_meta_ano = sorted(
    [col for col in meta_municipio.columns if col.startswith("meta_alfabetizacao_")]
 )

if not colunas_meta_ano:
    raise ValueError("Nenhuma coluna de meta por ano foi encontrada em meta_municipio.")

print(f"Colunas detectadas: {colunas_meta_ano}")
print("Etapa 4/9: transformando metas do formato largo para formato anual...")
id_vars = [
    col for col in ["id_municipio", "rede", "ano", "taxa_alfabetizacao"]
    if col in meta_municipio.columns
]

metas_por_ano_municipio = (
    meta_municipio
    .melt(
        id_vars=id_vars,
        value_vars=colunas_meta_ano,
        var_name="coluna_meta",
        value_name="meta_alfabetizacao"
    )
)

print("Etapa 5/9: extraindo o ano da meta e ajustando tipos...")
metas_por_ano_municipio["ano_meta"] = (
    metas_por_ano_municipio["coluna_meta"]
    .str.extract(r"(\d{4})", expand=False)
    .astype("Int64")
)
metas_por_ano_municipio["meta_alfabetizacao"] = pd.to_numeric(
    metas_por_ano_municipio["meta_alfabetizacao"], errors="coerce"
)

print("Etapa 6/9: limpando nulos e deixando apenas uma meta por municipio, rede e ano...")
metas_por_ano_municipio = (
    metas_por_ano_municipio
    .drop(columns=["coluna_meta"])
    .dropna(subset=["meta_alfabetizacao", "ano_meta"])
    .rename(columns={"taxa_alfabetizacao": "taxa_alfabetizacao_meta_base"})
    .drop_duplicates(subset=["id_municipio", "rede", "ano_meta"], keep="last")
    .sort_values(["id_municipio", "rede", "ano_meta"])
    .reset_index(drop=True)
)

print("Etapa 7/9: fazendo o join entre avaliacao e metas por municipio, rede e ano...")
avaliacao_para_merge = avaliacao_municipio.copy()
avaliacao_para_merge["ano_meta"] = avaliacao_para_merge["ano"]

silver_indicador_municipio = avaliacao_para_merge.merge(
    metas_por_ano_municipio[[
        "id_municipio",
        "rede",
        "ano_meta",
        "meta_alfabetizacao",
        "taxa_alfabetizacao_meta_base"
    ]],
    on=["id_municipio", "rede", "ano_meta"],
    how="left"
)

print("Etapa 8/9: adicionando metadados e salvando a tabela integrada...")
silver_indicador_municipio["_data_processamento"] = datetime.now()
silver_indicador_municipio["_camada_origem"] = "silver"

silver_indicador_municipio.to_parquet(
    SILVER_PATH / "silver_indicador_municipio.parquet",
    index=False
)

print("Etapa 9/9: validando a pergunta 'quero saber as metas de municipio por ano'...")
metas_municipio_por_ano = (
    silver_indicador_municipio[["id_municipio", "rede", "ano_meta", "meta_alfabetizacao"]]
    .dropna(subset=["meta_alfabetizacao"])
    .drop_duplicates()
    .sort_values(["ano_meta", "id_municipio", "rede"])
    .reset_index(drop=True)
)

print(f"Linhas avaliacao: {len(avaliacao_municipio):,}")
print(f"Linhas metas anuais: {len(metas_por_ano_municipio):,}")
print(f"Linhas tabela integrada: {len(silver_indicador_municipio):,}")
print("\nPercentual com meta encontrada:")
print((silver_indicador_municipio["meta_alfabetizacao"].notna().mean() * 100).round(2), "%")
print("\nQuantidade de metas por ano:")
print(metas_municipio_por_ano.groupby("ano_meta").size())

metas_municipio_por_ano.head(20)
```
