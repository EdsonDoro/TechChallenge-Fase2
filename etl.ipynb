**1. Configuração do Ambiente**

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import os

# Criando diretórios para organizar as camadas
os.makedirs('data/raw', exist_ok=True)
os.makedirs('data/bronze', exist_ok=True)
os.makedirs('data/silver', exist_ok=True)
os.makedirs('data/gold', exist_ok=True)

print("Ambiente configurado e pastas criadas!")
```

**2. Ingestão na Camada Bronze**

Ingestão das fontes de dados RAW para a camada bronze, em parquet, sem qualquer analise ou manupulação dos dados.

```python
def ingestao_bronze():
    # 1. Leitura do dado bruto (RAW)
    df = pd.read_csv('data/raw/br_inep_avaliacao_alfabetizacao_meta_alfabetizacao_brasil.csv')

    # 2. Adição de Metadados
    df['_data_ingestao'] = datetime.now()
    df['_origem'] = 'br_inep_avaliacao_alfabetizacao_meta_alfabetizacao_brasil_csv'

    # 3. Salvamento em formato Colunar (Parquet)
    caminho_bronze = 'data/bronze/meta_alfabetizacao_brasil.parquet'
    df.to_parquet(caminho_bronze, index=False)

    print(f"Ingestão Bronze concluída. Arquivo salvo em: {caminho_bronze}")
    return df

# Executando ingestão
df_bronze = ingestao_bronze()
df_bronze.shape
```
